## 📚 1. Теория: ключевые концепции

### 🔍 Object Detection: что оцениваем?

| Концепция | Формула/Описание | Зачем нужна |
|-----------|-----------------|-------------|
| **IoU** | `Intersection / Union` | Мера качества бокса |
| **NMS** | Подавление дубликатов по IoU-порогу | Убрать множественные детекции одного объекта |
| **Precision** | `TP / (TP + FP)` | Точность: какая доля детекций верна |
| **Recall** | `TP / (TP + FN)` | Полнота: какую долю объектов нашли |
| **AP** | Площадь под PR-кривой | Качество для одного класса |
| **mAP** | Среднее AP по классам/порогам | Итоговая метрика детектора |

### 🎯 Важное правило матчинга:
> **Один Ground Truth = максимум один True Positive**
> 
> Если модель предсказала 5 боксов на одного человека → только один станет TP (с лучшим IoU), остальные 4 → FP.

## 🔧 Часть 1: Non-Maximum Suppression (NMS)

### 📝 Задание 1.1: Реализация IoU
Напишите функцию, которая считает Intersection over Union для двух боксов в формате `[x1, y1, x2, y2]`.

```python
def calculate_iou(box1, box2):

    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    intersection = max(0, x2 - x1) * max(0, y2 - y1)
    area1 = (box1[2] - box1[0]) * (box1[3] - box1[1])
    area2 = (box2[2] - box2[0]) * (box2[3] - box2[1])
    union = area1 + area2 - intersection
    return intersection / union if union != 0 else 0


def test_iou():
    assert np.isclose(calculate_iou([0,0,10,10], [0,0,10,10]), 1.0), "Тест 1 failed"

    assert np.isclose(calculate_iou([0,0,10,10], [20,20,30,30]), 0.0), "Тест 2 failed"

    box_a = np.array([0, 0, 10, 10])
    box_b = np.array([5, 5, 15, 15])
    expected = 25 / (100 + 100 - 25)
    assert np.isclose(calculate_iou(box_a, box_b), expected, atol=1e-4), "Тест 3 failed"

    print("✅ Все тесты IoU пройдены!")

test_iou()
```

### 📝 Задание 1.2: Реализация NMS
Напишите функцию Non-Maximum Suppression, которая:
1. Сортирует боксы по уверенности (убывание)
2. Берет бокс с максимальной уверенностью → добавляет в результат
3. Удаляет все боксы, у которых IoU с выбранным > порога
4. Повторяет, пока боксы не кончатся

```python
def nms(boxes, scores, iou_threshold=0.5):
    if len(boxes) == 0:
        return []


    indices = sorted(range(len(boxes)), key=lambda i: scores[i])
    keep = []
    while indices:
        best = indices.pop()
        best_box = boxes[best]
        keep.append(best)
        new_indices = []
        for other in indices:
            other_box = boxes[other]
            if calculate_iou(best_box, other_box) < iou_threshold:
                new_indices.append(other)
        indices = new_indices

    return keep


def test_nms():
    test_boxes = np.array([
        [100, 100, 200, 200],
        [110, 110, 210, 210],
        [105, 105, 205, 205],
        [500, 500, 600, 600],
        [510, 510, 610, 610],
    ])
    test_scores = np.array([0.9, 0.8, 0.7, 0.95, 0.6])

    keep = nms(test_boxes, test_scores, iou_threshold=0.5)

    assert len(keep) == 2, f"Ожидали 2 бокса, получили {len(keep)}"

    kept_scores = test_scores[keep]
    assert set(kept_scores) == {0.95, 0.9}, f"Неверные скоры: {kept_scores}"

    print("✅ Все тесты NMS пройдены!")
    return test_boxes, test_scores, keep

boxes, scores, keep = test_nms()
```

## 📈 Часть 2: Mean Average Precision (mAP)

### 📝 Задание 2.1: Матчинг предсказаний с Ground Truth

```python
def match_predictions(gt_boxes, pred_boxes, pred_scores, iou_threshold=0.5):
    n_gt = len(gt_boxes)
    n_pred = len(pred_boxes)

    if n_pred == 0:
        return np.array([]), np.array([])

    tp_fp = np.zeros(n_pred)
    gt_matched = np.zeros(n_gt)


    pred_sorted = np.argsort(pred_scores)[::-1]
    sorted_scores = pred_scores[pred_sorted]
    sorted_boxes = pred_boxes[pred_sorted]

    for pred_id, pred_box in enumerate(sorted_boxes):
        best_match_ind, best_match_iou = None, None
        for gt_ind, gt_box in enumerate(gt_boxes):
            if not gt_matched[gt_ind]:
                curr_iou = calculate_iou(pred_box, gt_box)
                if curr_iou >= iou_threshold and (best_match_iou is None or curr_iou > best_match_iou):
                    best_match_iou = curr_iou
                    best_match_ind = gt_ind

        if best_match_ind is not None:
            gt_matched[best_match_ind] = 1
            tp_fp[pred_id] = 1
    return tp_fp, sorted_scores


def test_matching():
    np.random.seed(42)

    gt = np.array([
        [100, 100, 200, 200],
        [300, 300, 400, 400],
        [500, 100, 600, 200],
    ])

    pred = np.array([
        [105, 105, 205, 205, 0.9],
        [110, 110, 210, 210, 0.8],
        [305, 305, 405, 405, 0.95],
        [50, 50, 80, 80, 0.99],
        [500, 100, 600, 200, 0.7],
    ])

    tp_fp, scores = match_predictions(gt, pred[:, :4], pred[:, 4], iou_threshold=0.5)

    expected_tp_fp = np.array([0, 1, 1, 0, 1])

    assert np.array_equal(tp_fp, expected_tp_fp), f"Expected {expected_tp_fp}, got {tp_fp}"
    assert np.all(scores == np.sort(pred[:, 4])[::-1]), "Scores не отсортированы!"

    print("✅ Тест матчинга пройден!")
    return gt, pred, tp_fp, scores

gt, pred, tp_fp, scores = test_matching()
```

### 📝 Задание 2.2: Precision-Recall Curve и AP

```python
def calculate_precision_recall(tp_fp, n_gt_total):
    n_pred = len(tp_fp)
    if n_pred == 0:
        return np.array([]), np.array([])

    cum_tp = np.cumsum(tp_fp)
    cum_fp = np.cumsum(1 - tp_fp)


    precision = cum_tp / (cum_tp + cum_fp + 1e-6)
    recall = cum_tp / n_gt_total if n_gt_total != 0 else np.zeros_like(cum_tp)
    return precision, recall


def calculate_ap_coco(precision, recall):
    if len(precision) == 0:
        return 0.0

    mrec = np.concatenate(([0.], recall, [1.]))
    mpre = np.concatenate(([0.], precision, [0.]))

    for i in range(len(mpre) - 2, -1, -1):
        mpre[i] = max(mpre[i], mpre[i + 1])

    i = np.where(mrec[1:] != mrec[:-1])[0]

    ap = np.sum((mrec[i + 1] - mrec[i]) * mpre[i + 1])
    return ap


def test_pr_curve():
    tp_fp = np.array([0, 1, 1, 0, 1])
    n_gt = 3

    prec, rec = calculate_precision_recall(tp_fp, n_gt)
    ap = calculate_ap_coco(prec, rec)

    print(f"Precision: {prec}")
    print(f"Recall: {rec}")
    print(f"AP@0.5: {ap:.4f}")

    plt.figure(figsize=(8, 6))
    plt.plot(rec, prec, marker='o', linewidth=2, markersize=6)
    plt.xlabel('Recall', fontsize=12)
    plt.ylabel('Precision', fontsize=12)
    plt.title(f'Precision-Recall Curve (AP@0.5 = {ap:.3f})', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.xlim([0, 1.05])
    plt.ylim([0, 1.05])
    plt.show()

    return ap

ap_val = test_pr_curve()
```

### 📝 Задание 2.3: mAP@0.5:0.95 (COCO Standard)

```python
def calculate_map_coco(gt_boxes, pred_boxes, pred_scores, iou_thresholds=None):
    if iou_thresholds is None:
        iou_thresholds = np.arange(0.5, 0.96, 0.05)

    aps = []

    for thresh in iou_thresholds:
        tp_fp, scores = match_predictions(gt_boxes, pred_boxes, pred_scores, iou_threshold=thresh)
        if len(tp_fp) == 0:
            aps.append(0.0)
            continue
        prec, rec = calculate_precision_recall(tp_fp, len(gt_boxes))
        ap = calculate_ap_coco(prec, rec)
        aps.append(ap)

    return np.mean(aps), np.array(aps), iou_thresholds


def full_test_map():
    np.random.seed(123)

    gt = np.array([
        [100, 100, 200, 200],
        [300, 300, 400, 400],
        [500, 100, 600, 200],
        [100, 500, 200, 600],
        [800, 800, 900, 900]
    ])

    pred = np.array([
        [105, 105, 205, 205, 0.95],
        [110, 110, 210, 210, 0.80],
        [305, 305, 405, 405, 0.85],
        [500, 100, 600, 200, 0.90],
        [100, 500, 200, 600, 0.40],
        [900, 900, 950, 950, 0.88],
        [50, 50, 80, 80, 0.99],
        [805, 805, 895, 895, 0.70]
    ])

    mAP, aps, thresholds = calculate_map_coco(gt, pred[:, :4], pred[:, 4])

    print(f"📊 Результаты:")
    print(f"mAP@0.5:0.95 = {mAP:.4f}")
    print(f"\nAP по порогам:")
    for t, ap in zip(thresholds, aps):
        print(f"  IoU={t:.2f} → AP={ap:.4f}")

    plt.figure(figsize=(10, 4))
    plt.plot(thresholds, aps, marker='s', linewidth=2)
    plt.axhline(y=mAP, color='r', linestyle='--', label=f'mAP = {mAP:.3f}')
    plt.xlabel('IoU Threshold', fontsize=12)
    plt.ylabel('AP', fontsize=12)
    plt.title('AP при разных порогах IoU (COCO-style)', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

    return mAP

mAP_result = full_test_map()
```

### NMS Vectorized

```python
def nms_vector(boxes, scores, iou_threshold=0.5):
    if len(boxes) == 0:
        return []

    indices = np.argsort(scores)[::-1]
    keep = []

    while indices.size > 0:
        keep.append(indices[0])
        if len(indices) == 1:
            break

        selected_box = boxes[indices[0]]
        other_boxes = boxes[indices[1:]]

        xx1 = np.maximum(selected_box[0], other_boxes[:, 0])
        yy1 = np.maximum(selected_box[1], other_boxes[:, 1])
        xx2 = np.minimum(selected_box[2], other_boxes[:, 2])
        yy2 = np.minimum(selected_box[3], other_boxes[:, 3])

        w = np.maximum(0, xx2 - xx1)
        h = np.maximum(0, yy2 - yy1)
        inter = w * h

        area_sel = (selected_box[2] - selected_box[0]) * (selected_box[3] - selected_box[1])
        area_oth = (other_boxes[:, 2] - other_boxes[:, 0]) * (other_boxes[:, 3] - other_boxes[:, 1])
        union = area_sel + area_oth - inter

        iou = inter / union

        valid_mask = iou < iou_threshold
        indices = indices[1:][valid_mask]

    return keep

def test_nms_vector():
    test_boxes = np.array([
        [100, 100, 200, 200],
        [110, 110, 210, 210],
        [105, 105, 205, 205],
        [500, 500, 600, 600],
        [510, 510, 610, 610],
    ])
    test_scores = np.array([0.9, 0.8, 0.7, 0.95, 0.6])

    keep = nms_vector(test_boxes, test_scores, iou_threshold=0.5)
    kept_scores = test_scores[keep]

    assert len(keep) == 2, f"Ожидали 2 бокса, получили {len(keep)}"
    assert set(kept_scores) == {0.95, 0.9}, f"Неверные скоры: {kept_scores}"
    print("✅ Тест векторизованного NMS пройден!")

test_nms_vector()
```

### Multiclass Matching

```python
def match_predictions_multiclass(gt_boxes, gt_classes, pred_boxes, pred_scores, pred_classes, iou_threshold=0.5):
    all_classes = np.unique(np.concatenate([gt_classes, pred_classes]))
    results = {}

    for cls in all_classes:
        gt_mask = gt_classes == cls
        pred_mask = pred_classes == cls

        cls_gt = gt_boxes[gt_mask]
        cls_pred = pred_boxes[pred_mask]
        cls_scores = pred_scores[pred_mask]

        tp_fp, sorted_scores = match_predictions(cls_gt, cls_pred, cls_scores, iou_threshold)
        results[int(cls)] = (tp_fp, sorted_scores, int(np.sum(gt_mask)))

    return results

def test_matching_multiclass():
    np.random.seed(42)
    gt_boxes = np.array([[10, 10, 50, 50], [200, 200, 250, 250]])
    gt_classes = np.array([0, 1])

    pred_boxes = np.array([
        [12, 12, 52, 52], [15, 15, 55, 55],
        [205, 205, 255, 255]
    ])
    pred_scores = np.array([0.9, 0.8, 0.85])
    pred_classes = np.array([0, 0, 1])

    res = match_predictions_multiclass(gt_boxes, gt_classes, pred_boxes, pred_scores, pred_classes, iou_threshold=0.5)

    assert 0 in res and 1 in res, "Отсутствуют ключи классов"
    assert len(res[0][0]) == 2, f"Ожидали 2 предсказания для класса 0, получили {len(res[0][0])}"
    assert res[0][0][0] == 1 and res[0][0][1] == 0, "TP/FP для класса 0 неверны"
    assert res[1][0][0] == 1, "TP для класса 1 неверен"

    print("✅ Тест мультиклассового матчинга пройден!")

test_matching_multiclass()
```

### Sensitivity Analysis

### Soft-NMS

```python
def soft_nms(boxes, scores, iou_threshold=0.5, sigma=0.5, conf_thresh=0.001):
    if len(boxes) == 0:
        return []

    boxes = np.array(boxes)
    scores = np.array(scores, dtype=float)
    keep = []
    indices = np.arange(len(scores))

    while indices.size > 0:
        max_idx = indices[np.argmax(scores[indices])]
        keep.append(max_idx)

        indices = indices[indices != max_idx]
        if indices.size == 0:
            break

        selected_box = boxes[max_idx]
        other_boxes = boxes[indices]

        xx1 = np.maximum(selected_box[0], other_boxes[:, 0])
        yy1 = np.maximum(selected_box[1], other_boxes[:, 1])
        xx2 = np.minimum(selected_box[2], other_boxes[:, 2])
        yy2 = np.minimum(selected_box[3], other_boxes[:, 3])
        w = np.maximum(0, xx2 - xx1)
        h = np.maximum(0, yy2 - yy1)
        inter = w * h

        area_sel = (selected_box[2]-selected_box[0])*(selected_box[3]-selected_box[1])
        area_oth = (other_boxes[:,2]-other_boxes[:,0])*(other_boxes[:,3]-other_boxes[:,1])
        iou = inter / (area_sel + area_oth - inter + 1e-6)

        decay_mask = iou > iou_threshold
        scores[indices] *= np.where(decay_mask, np.exp(-(iou**2) / sigma), 1.0)

    result = [idx for idx in keep if scores[idx] > conf_thresh]
    return result, scores[result]

def test_soft_nms():
    test_boxes = np.array([
        [100, 100, 200, 200],
        [110, 110, 210, 210],
        [105, 105, 205, 205],
        [500, 500, 600, 600],
    ])
    test_scores = np.array([0.9, 0.8, 0.7, 0.95])

    keep, final_scores = soft_nms(test_boxes, test_scores, iou_threshold=0.5, sigma=0.5, conf_thresh=0.1)

    print(f"✅ Soft-NMS тест пройден! Оставлено боксов: {len(keep)}")
    print(f"📉 Итоговые confidence: {final_scores.round(3)}")

test_soft_nms()
```

### 🐛 Частые ошибки:
1. **IoU = 0 при касании:** Забыли `max(0, x2-x1)` → отрицательная площадь пересечения ❌
2. **Деление на ноль:** В Precision, когда `TP+FP = 0` → добавьте `+1e-6` в знаменатель
3. **Порядок сортировки:** Забыли сортировать по confidence перед матчингом → кривая PR будет неверной
4. **GT Matching:** Один GT помечают как matched для нескольких предсказаний → нарушается правило "один объект — одна детекция"
